### 3. Array Operations:

● Convert the temperatures_w1 from Celsius to Fahrenheit using the
formula: 

Fahrenheit = (Celsius * 9/5) + 32

```python
#creating a function that accepts an array and returns a new array

def convert_temperature(temperatures_w1):
   temperatures_w1_F = (temperatures_w1 * 9/5) + 32     #assigning the formula of conversion to a new variable for temperature in Farenheit
   return  temperatures_w1_F

temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

result = convert_temperature(temperatures_w1)

print(result)
```

● Find the maximum, minimum, and mean temperatures for the week.

```python
# Maximum temperature for the week 

max=np.max(temperatures_w1)
print(f"temperatures_w1:{temperatures_w1}")
print(f" Maximum temperature of the week:{max}")
```

```python
# Minimum temperature for the week 

min=np.min(temperatures_w1)
print(f"temperatures_w1:{temperatures_w1}")
print(f"Minimum temperature of the week:{min}")
```

```python
# mean temperature for the week

mean=np.mean(temperatures_w1)

# Rounding the mean value to two decimal places

rounded_mean = np.round(mean, 2)


print(f"temperatures_w1:{temperatures_w1}")
print(f"Mean temperature of the week:{rounded_mean}")
```

### 4. Array Slicing and Indexing:

● Extract the temperatures for the first three days of the week.

```python
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Extract temperatures for the first three days

first_three_days = temperatures_w1[:3]    # Slicing the array to get first three elements

print(f"temperatures_w1:{temperatures_w1}")
print(f"Temperature of first three days:{first_three_days}")
```

● Extract the temperatures for the weekend (last two days).

```python
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Extract temperatures for the last two days

last_two_days = temperatures_w1[-2:]    # Slicing the array to get last two elements

print(f"temperatures_w1:{temperatures_w1}")
print(f"Temperature of last two days:{last_two_days}")
```

● Extract the temperatures for the middle three days of the week.


```python
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Extract temperatures for the middle three days of the week.

middle_three_days = temperatures_w1[2:5]    # Slicing the array to get middle three elements

print(f"temperatures_w1:{temperatures_w1}")
print(f"Temperature of middle three days:{middle_three_days}")
```

### 1. Creating Pandas Series:

● Create a Pandas Series named marks with the following data:

➢ Marks: 95, 92, 89, 85, 80

➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

```python
# remove future warning error
import warnings
warnings.filterwarnings("ignore")
```

```python
# Creatinga Pandas Series named marks 

Marks=pd.Series(data=[95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

print(Marks)
```

### 2. Indexing and Slicing:

● Use the integer index position to access the mark of the 1st rank student.

```python
#accessing the mark of rank 1 student

print("Mark of rank 1 student:",Marks.iloc[0])
```

● Use the loc accessor to retrieve the marks of the top 3 ranks by
specifying their index labels.

```python
#Retrieving the top 3 ranks using loc

print(Marks.loc[['Rank1', 'Rank2', 'Rank3']])

#loc is used to access values using their index labels.
```

● Use the iloc accessor to retrieve the mark of the 3rd rank student.

```python
#Using the iloc accessor to retrieve the mark of the 3rd rank student.

print("Mark of third rank student:",Marks.iloc[2])   #iloc accesses values using integer positions.
```

● Apply a boolean mask to filter the Series and retrieve the ranks where the
marks are greater than 90.

```python


print(Marks)
print()
print(Marks>90)                #creates a Boolean mask.
result = Marks[Marks > 90]     #filters the Series using that mask
print()
print("Marks greater than 90 :",result)
```

### 3. Manipulating Series:

● Modify the mark of the student with the 1st rank to 100.

```python
Marks['Rank1'] = 100

print(Marks)
```

● Remove the entry corresponding to the last rank from the Series.

```python
#drop() removes the specified index.

#We assign the result back to 'Marks' to update the original variable.

Marks = Marks.drop('Rank5')

print(Marks)
```

● Compute the CGPA by dividing each mark in the Series by 10.

```python
# Each mark is divided by 10 to get the cgpa

cgpa = Marks / 10

print(cgpa)
```

## Pandas DataFrame:

### 1. Creating Pandas DataFrame

● Create a Pandas DataFrame named transactions with the following data:

➢ TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110

➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    
➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
'South', 'North'

➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

```python
#creaating a dataframe

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing','Furniture', 'Electronics' ],
    'Region': [ 'North', 'South', 'North', 'East', 'West','North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

#Dictionary keys become column names.Dictionary values become column data.Each row represents one transaction.


transactions = pd.DataFrame(data)   #pd.DataFrame() converts a dictionary into a DataFrame.
transactions
```

● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200

```python
#We are using two conditions:
#1. transactions['Region'] == 'North' checks whether the region is North.
#2. transactions['Amount'] > 200 checks whether the amount is greater than 200.
#3. & means AND. Both conditions must be True

filtered_data = transactions[(transactions['Region'] == 'North') &(transactions['Amount'] > 200)]

filtered_data

```
